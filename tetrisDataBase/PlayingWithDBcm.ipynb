## Sqlite3

```python
import DBcm

SQL = """
    create table if not exists bond (
        name varchar(32) not null 
    )
"""

names = [
    ("Sean Connery",),
    ("George Lazenby",),
    ("Roger Moore",),
    ("Timothy Dalton",),
    ("Pierse Brosnan",),
    ("Daniel Craig",),
    ("Robert Pattinson",),
]

SQL2 = """
    insert into bond
    (name)
    values
    (?)
"""

SQL3 = """
    select * from bond
"""
```

```python
with DBcm.UseDatabase( "bond.sqlite" ) as db:
    db.execute(SQL)
```

```python
with DBcm.UseDatabase( "bond.sqlite" ) as db:
    db.executemany(SQL2, names)
```

```python
with DBcm.UseDatabase( "bond.sqlite" ) as db:
    db.execute(SQL3)
    results = db.fetchall()
results
```
